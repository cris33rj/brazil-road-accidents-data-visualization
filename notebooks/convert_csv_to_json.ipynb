# Convert PRF accident data from CSV to JSON

Converts the consolidated PRF dataset (`docs/data/datatran.csv`, `;`-separated) into a JSON array of records.

```python
from pathlib import Path

import pandas as pd

DATA_DIR = Path("../docs/data")
OUTPUT_DIR = Path("output")
OUTPUT_DIR.mkdir(exist_ok=True)

data = pd.read_csv(DATA_DIR / "datatran.csv", sep=";", low_memory=False)

json_file = OUTPUT_DIR / "datatran.json"
data.to_json(json_file, orient="records", force_ascii=False)

print(f"{len(data)} records written to {json_file}")
```
