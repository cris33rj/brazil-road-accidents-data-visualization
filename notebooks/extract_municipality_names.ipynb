# Extract municipality IDs and names from GeoJSON

Reads the municipality polygons (`docs/data/brasil_completo.json`), normalizes names (uppercase, no accents) so they can be joined with the `municipio` field of the PRF dataset, and saves an `id,name` table.

```python
import json
from pathlib import Path

import pandas as pd
from unidecode import unidecode

DATA_DIR = Path("../docs/data")
OUTPUT_DIR = Path("output")
OUTPUT_DIR.mkdir(exist_ok=True)

with open(DATA_DIR / "brasil_completo.json", encoding="utf-8") as file:
    data = json.load(file)

rows = [
    {"id": feature["properties"]["id"], "name": unidecode(feature["properties"]["name"]).upper()}
    for feature in data["features"]
]
df = pd.DataFrame(rows)

csv_file = OUTPUT_DIR / "municipios_ids_e_nomes.csv"
df.to_csv(csv_file, index=False, encoding="utf-8")

print(f"{len(df)} municipalities written to {csv_file}")
```
